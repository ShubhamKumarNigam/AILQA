```python
import pandas as pd
import numpy as np
import csv
import os,glob
```

```python
def readfiles():
    dr=os.listdir('')
    return dr
```

```python
judgements=(readfiles())
for i in range(len(judgements)):
    judgements[i]=''+judgements[i]
```

```python
csv_files=[]
txt_files=[]
for i in range(len(judgements)):
    if '.txt' in judgements[i]:
        txt_files.append(judgements[i])
    else:
        csv_files.append(judgements[i])
```

## Data Dumping

### Text Dump

```python
itr=0
for i in range(len(txt_files)):
    data=(open(txt_files[i],'r')).read()
    f=open(''+'doc_'+str(itr)+'.txt','w+')
    f.write(data)
    f.close()
    itr+=1
```

### CSV Dump


```python
for i in range(len(csv_files)):
    data=pd.read_csv(csv_files[i])
    for j in range(len(data)):
        txt='Title:\n'+data['title'][j]+'\n'+'Content:\n'+data['content'][j]
        f=open(''+'doc_'+str(itr)+'.txt','w+')
        f.write(txt)
        f.close()
        itr+=1
```
