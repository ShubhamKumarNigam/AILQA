```python
import pandas as pd
import numpy as np
import csv
import os,glob
```

```python
def readfiles():
    dr=os.listdir('')
    return dr
```

```python
judgements=(readfiles())
for i in range(len(judgements)):
    judgements[i]=''+judgements[i]
```

```python
crime_judgements=[]
head_note=[]
for i in range(len(judgements)):
    doc=open(judgements[i],'r').read()
    if 'CRIMINAL' in doc:
        crime_judgements.append(doc)
    if 'HEADNOTE' in doc:
        head_note.append(doc)
```

```python
for i in range(len(crime_judgements)):
    f=open(''+'crime_judgement_'+str(i)+'.txt','w+')
    f.write(crime_judgements[i])
    f.close()
```
