```python
import pandas as pd
import numpy as np
```

```python
from PyPDF2 import PdfReader
import glob, os
def readfiles():
    os.chdir('')
    pdfs=[]
    for file in glob.glob("*.pdf"):
        pdfs.append(file)
    return pdfs
```

```python
pdfs=readfiles()
for i in range(len(pdfs)):
    pdfs[i]=''+pdfs[i]
```

```python
for path in pdfs:
    reader = PdfReader(path)
    pages = reader.pages
    text=''
    for page in pages:
        text+=page.extract_text()
    f=open(path.rstrip('.pdf')+'.txt','w+')
    f.write(text)
    f.close()
```
