```python
import torch
from transformers import AutoTokenizer
import pandas as pd
import numpy as np
import csv
from tqdm import tqdm
import os,glob
```

```python
def readfiles():
    os.chdir('')
    doc=[]
    for file in glob.glob("*.txt"):
        doc.append(file)
    return doc
```

```python
judgements=(readfiles())
for i in range(len(judgements)):
    judgements[i]=''+judgements[i]
```

```python
txt=''
for i in tqdm(range(len(judgements))):
    f=open(judgements[i],'r')
    data=f.read()
    txt=txt+data+'\n'
txt=txt.strip()
```

```python
tokenizer = AutoTokenizer.from_pretrained("gpt2")
```

```python
input_ids = torch.tensor(tokenizer.encode(txt)).unsqueeze(0)
num_tokens = input_ids.shape[1]
print(num_tokens)
```
