```python
import pandas as pd
import numpy as np
import os
```

```python
def calculate_average_scores(directory):
    # Create an empty list to store average scores
    avg_scores_list = []
    
    for filename in os.listdir(directory):
        if filename.endswith('.csv'):
            filepath = os.path.join(directory, filename)
            
            # Read the CSV file into a DataFrame
            df = pd.read_csv(filepath)
            # Calculate average scores for each column
            avg_scores = {
                'filename': filename.rstrip('.csv'),
                'rouge_1_avg': round(df['rouge_1'].mean(), 2),
                'rouge_2_avg': round(df['rouge_2'].mean(), 2),
                'rouge_l_avg': round(df['rouge_l'].mean(), 2),
                'meteor_score_avg': round(df['meteor_score'].mean(), 2),
                'similarity_avg': round(df['similarity'].mean(), 2),
                'score_avg': round(df['score'].mean(), 2)
            }
            
            # Append average scores to the list
            avg_scores_list.append(avg_scores)
    
    # Convert the list of dictionaries to a DataFrame
    avg_scores_df = pd.DataFrame(avg_scores_list, columns=['filename', 'rouge_1_avg', 'rouge_2_avg', 'rouge_l_avg', 'meteor_score_avg', 'similarity_avg', 'score_avg'])
    
    return avg_scores_df
```

```python
def count_score_values(directory):
    # Create an empty list to store score counts
    score_counts_list = []
    
    for filename in os.listdir(directory):
        if filename.endswith('.csv'):
            filepath = os.path.join(directory, filename)
            
            # Read the CSV file into a DataFrame
            df = pd.read_csv(filepath)
            # Count occurrences of each score value
            score_counts = {
                'filename': filename,
                '1': (df['score'] == 1).sum(),
                '2': (df['score'] == 2).sum(),
                '3': (df['score'] == 3).sum(),
                '4': (df['score'] == 4).sum(),
                '5': (df['score'] == 5).sum()
            }
            
            # Append score counts to the list
            score_counts_list.append(score_counts)
    
    # Convert the list of dictionaries to a DataFrame
    score_counts_df = pd.DataFrame(score_counts_list, columns=['filename', '1', '2', '3', '4', '5'])
    
    return score_counts_df
```

```python
# Call the function to calculate average scores
average_scores_dataframe = calculate_average_scores(directory)

```

```python
def count_scores_by_type(directory):
    # Create an empty list to store score counts by type
    score_counts_by_type_list = []
    
    for filename in os.listdir(directory):
        if filename.endswith('.csv'):
            filepath = os.path.join(directory, filename)
            
            # Read the CSV file into a DataFrame
            df = pd.read_csv(filepath)
            
            # Group by 'type' and count the occurrences of each score for each type
            grouped = df.groupby('type')['score'].value_counts().unstack(fill_value=0).reset_index()
            grouped['filename'] = filename.rstrip('.csv')
            
            # Append each row to the list
            for _, row in grouped.iterrows():
                score_counts = {
                    'filename': row['filename'],
                    'type': row['type'],
                    'score_1': row.get(1, 0),
                    'score_2': row.get(2, 0),
                    'score_3': row.get(3, 0),
                    'score_4': row.get(4, 0),
                    'score_5': row.get(5, 0)
                }
                score_counts_by_type_list.append(score_counts)
    
    # Convert the list of dictionaries to a DataFrame
    score_counts_by_type_df = pd.DataFrame(score_counts_by_type_list, columns=['filename', 'type', 'score_1', 'score_2', 'score_3', 'score_4', 'score_5'])
    
    return score_counts_by_type_df
```

```python
# Provide the directory containing the CSV files
directory = ''
```

```python
score_by_type=count_scores_by_type(directory)
```
