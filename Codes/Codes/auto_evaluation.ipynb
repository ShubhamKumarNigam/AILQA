```python
from langchain.evaluation import load_evaluator
from langchain_openai import ChatOpenAI
from tqdm import tqdm
```

## Load Test Data

```python
path=''
data=pd.read_csv(path)
```

```python
data.head(2)
```

```python
evaluator = load_evaluator("qa",llm=ChatOpenAI(model="gpt-4"))
```

```python
scores=[]
for _, row in tqdm(data.iterrows()):

    input=row['question'].strip()
    reference=row['answer'].strip()
    prediction=row['answer_generated']

    score=evaluator.evaluate_strings(prediction=prediction,input=input,reference=reference)

    scores.append(score)
```

```python
data['gpt_evaluation']=scores
```

```python
data.to_csv(path,index=False)
```
