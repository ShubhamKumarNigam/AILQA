## PLOT FOR QUESTION TYPE SCORE

```python
def superimpose_histograms(df1, df2, df1_name, df2_name, plot_save_name):
    # Ensure columns are named 1, 2, 3, 4, 5 and 'type'
    columns = ['1', '2', '3', '4', '5']
    
    # Check if the dataframes have the required columns
    required_columns = columns + ['type']
    if not all(col in df1.columns for col in required_columns) or not all(col in df2.columns for col in required_columns):
        raise ValueError("Both dataframes must have columns named 1, 2, 3, 4, 5 and 'type'")
    
    # Get unique types
    types = pd.concat([df1['type'], df2['type']]).unique()
    
    # Determine the layout of the subplots
    num_types = len(types)
    num_cols = 2  # You can adjust this to have more columns
    num_rows = (num_types + num_cols - 1) // num_cols
    
    fig, axes = plt.subplots(num_rows, num_cols, figsize=(15, num_rows * 5))
    axes = axes.flatten()
    
    # Plot histograms for each type
    for i, t in enumerate(types):
        df1_type = df1[df1['type'] == t]
        df2_type = df2[df2['type'] == t]
        
        if df1_type.empty or df2_type.empty:
            continue
        
        means_df1 = df1_type[columns].mean()
        means_df2 = df2_type[columns].mean()
        
        x = range(1, 6)
        
        axes[i].bar(x, means_df1, width=0.4, alpha=0.5, label=df1_name, align='center')
        axes[i].bar(x, means_df2, width=0.4, alpha=0.5, label=df2_name, align='edge')
        
        axes[i].set_title(f'Number of Question VS Score Plot For {t}')
        axes[i].set_xlabel('Score')
        axes[i].set_ylabel('Number of Question')
        axes[i].set_xticks(x)
        axes[i].set_xticklabels(columns)
        axes[i].legend(loc='upper left', bbox_to_anchor=(1, 1))
    
    # Remove unused subplots
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])
    
    plt.tight_layout(rect=[0, 0, 0.85, 1])
    plt.savefig(f'Output with mxbai/Plots/{plot_save_name}.pdf', format='pdf')
    plt.show()
```

```python
superimpose_histograms(data1,data2,'GPT-3.5-Turbo','GPT-3.5-Turbo_WITH_RAG','question_type_plot_for_gpt')
```

## PLOT FOR AUTO EVALUATION SCORE

```python
listdir=os.listdir('')
```

```python
rows=[]
for dir in listdir:
    filepath=f"/{dir}"
    data=pd.read_csv(filepath)
    data['gpt_evaluation']=data['gpt_evaluation'].apply(lambda x: ast.literal_eval(x))
    score=data['gpt_evaluation'].apply(lambda x: x['score'])
    rows.append([dir.rstrip('.csv').rstrip('_output'),list(score)])
```

```python
df=pd.DataFrame(rows,columns=['filename','scores'])
```

```python
# Count the number of 1 scores in each file
df['count_ones'] = df['scores'].apply(lambda x: x.count(1))
```

```python
df
```

```python
df['filename'][0]='LLAMA3-70b'
df['filename'][1]='LLAMA3-70b_WITH_RAG'
df['filename'][2]='MIXTRAL-8x7b'
df['filename'][3]='MIXTRAL-8x7b_WITH_RAG'
df['filename'][4]='GPT-3.5-Turbo'
df['filename'][5]='GPT-3.5-Turbo_WITH_RAG'
```

```python
# Sort the DataFrame by the count of 1 scores in descending order
df = df.sort_values(by='count_ones', ascending=True)
```

```python
# Create the horizontal bar plot
plt.figure(figsize=(10, 6))
plt.barh(df['filename'], df['count_ones'], color='skyblue', edgecolor='black')
plt.xlabel('AUTO EVALUATED SCORE:1')
plt.ylabel('MODEL')
# plt.title('Horizontal Bar Plot of 1 Scores Count for Each File')
plt.grid(axis='x')

# Save the plot as a PDF file
plt.savefig('', format='pdf')

# Display the plot
plt.show()
```
