```python
import pandas as pd
import scipy.stats as stats

import numpy as np
from scipy import stats
# from statsmodels.stats.multicomp import pairwise_tukeyhsd, MultiComparison
import pandas as pd
```

```python
base_path = ""
```

```python
file_names = [
    'Davinci.csv',
    'LLama2-70b.csv',
    'Ada+Davinci.csv',
    'Instructor+Davinci.csv',
    'BM25+Davinci.csv',
    'Ada+LLama2-70b.csv',
    'Instructor+LLama2-70b.csv',
    'Ada+Flan-UL2.csv',
    'Instructor+Flan-UL2.csv'
]
```

```python
mpnet_score_dict ={}
mpnet_scores_arrays = {}
for files in file_names:
    df = pd.read_csv(base_path+files, usecols=['mpnet_score'])  # Read only the mpnet_score column
    file_name = files.split('/')[-1][:-4]  # Extracting file name from the path
    print(file_name)
    mpnet_score_dict[file_name] = df
#     print(df)
    mpnet_scores_arrays[file_name + "_scores"] = np.array(df['mpnet_score'])
```

```python
df = pd.DataFrame(mpnet_scores_arrays)

# Initialize a DataFrame for p-values
p_values = pd.DataFrame(index=df.columns, columns=df.columns)

# Calculate p-values
for model1 in df.columns:
    for model2 in df.columns:
        if model1 != model2:
            _, p_value = stats.ttest_rel(df[model1], df[model2])
            p_values.at[model1, model2] = p_value

# Mask upper triangle of the DataFrame
mask = np.tril(np.ones_like(p_values, dtype=bool))
lower_tri_p_values = p_values.where(mask)

# Optional: Adjust for multiple comparisons
adjusted_p_values = lower_tri_p_values.apply(lambda x: x * (len(df.columns) - 1))
adjusted_p_values = adjusted_p_values.clip(upper=1.0)

# Round to four decimal places and replace NaN with '-'
lower_tri_p_values = lower_tri_p_values.round(4).fillna('-')
adjusted_p_values = adjusted_p_values.round(4).fillna('-')

# Display lower triangular p-values
# print("Lower Triangular P-Values:\n", lower_tri_p_values)
# print("\nLower Triangular Adjusted P-Values:\n", adjusted_p_values)
```

```python
pd.options.display.float_format = '{:.4f}'.format
```
